```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('gambar.jpg',cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray')
plt.title('Citra Asli (Grayscale)')
plt.show()

Binary =cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)[1]
plt.imshow(Binary, cmap='gray')
plt.title('Citra Bineri')
plt.show()
```

```python
negative = 255 - img
plt.imshow(negative,cmap='gray')
plt.title('citra Negative')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Baca gambar dalam mode grayscale
img = cv2.imread('gambar.jpg', cv2.IMREAD_GRAYSCALE)

# Cek apakah gambar terbaca
if img is None:
    raise FileNotFoundError("❌ Gambar tidak ditemukan! Pastikan path benar.")

# Transformasi logaritmik
c = 255 / np.log(1 + np.max(img))
log_image = c * np.log(1 + img)
log_image = np.array(log_image, dtype=np.uint8)

# Tampilkan hasil
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('Citra Asli')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(log_image, cmap='gray')
plt.title('Logarithmic Transformation')
plt.axis('off')

plt.show()
```

```python
gamma = 0.5
c = 1
power_law = c * np.power(img / 255.0, gamma)
power_law = np.uint8(power_law * 255)

plt.imshow(power_law, cmap='gray')
plt.title(f'Power-law')
plt.show
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('gambar.jpg',cv2.IMREAD_GRAYSCALE)

Kernel = np.array([[-0,-1,-0],[-1,5,-1],[0,-1,0]])

conv = cv2.filter2D(img,-1,Kernel)

plt.subplot(1,2,1), plt.imshow(img,cmap='gray'),plt.title('Asli.')
plt.subplot(1,2,2), plt.imshow(conv,cmap='gray'),plt.title('Hasil.')
plt.show()
```
